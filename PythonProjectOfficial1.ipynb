```python
#Data Cleaning
df.isnull().sum()
```

```python
print('Missing Values:')
print(df.isna().sum())
```

```python
#Convert Mean Score to Numeric (float) and values that is a letter becomes NaN
df['Mean Score'] = pd.to_numeric(df['Mean Score'], errors='coerce')
```

```python
print('Missing Values:', df['Mean Score'].isna().sum())
```

```python
#Initial Comparison Point: Francis Lewis High School
school_name = 'Francis Lewis High School'
School_df = df[df['School Name'] == school_name]
```

```python
#Descriptive Statistics of Francis Lewis High School
School_df['Mean Score'].describe()
```

```python
School_df['Mean Score'].agg(['median'])
```

```python
Other_school_df = df[df['School Name'] != school_name]
```

```python
#Descriptive Statistics for all schools
Other_school_df['Mean Score'].describe()
```

```python
Other_school_df['Mean Score'].agg(['median'])
```

```python
#Compare
School_Average = School_df['Mean Score'].mean()
Other_School_Average = Other_school_df['Mean Score'].mean()

print('Francis Lewis High School:', School_Average)
print('Other Schools:', Other_School_Average)
```

```python
#Visualization
plt.bar(['Francis Lewis High School', 'Other Schools'], [School_Average, Other_School_Average], color=['blue'])
plt.title('Average Mean Score: Francis Lewis High School vs Other Schools')
plt.ylabel('Average Mean Score')
plt.xlabel('School(s)')
plt.xticks(rotation=45)
plt.show()
```

```python
#Conclusion: The average mean score of Francis Lewis High School is compared to the average mean score of other schools.
Descriptive statistics and bar plot were used, showing that compared to other schools, Francis Lewis High School has a higher average mean score.
```
